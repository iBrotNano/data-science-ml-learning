## Count Words

With the `Counter` class from the `collections` module you can easily count the number of occurrences of each word in a text. This can be useful for various applications, such as analyzing the frequency of words in a document or creating a word cloud.

```python
from collections import Counter

text = """Fluffum schnipsum wackelito banana-dolor sit amet, kicher elit.
Pinguinibus hüpfendibus quantum quatschum, spaghetti tempor blubb.
Kartoffel maximus ploppity pling, senf und salsa con gaudi.
Nudeltronic wummsi deluxe, ipsum klabauter randomicus.
Miau miau, burrito boom, zebrastreifen mit Glitzerhelm.
Taco-loco schnurpsel di doo, pling-plong philosophicus.
Und am Ende sagt der Toaster: „Ich bin bereit für den Weltfrieden und Marmelade."""

parts = text.split() # splits the text on spaces
counter = Counter(parts)
print(counter.items())
```

## Replace Substrings

There are some methods to replace subtrings in a string.

```python
text = "Hello"
text = text.replace("Hello", "Hi")
print(text)

text = """Die Kräfte knallen, die Fäden verflechten sich, ein Klangteppich voller Funkelfragmente entsteht.
Man könnte fast glauben, ein Kakadu führe ein Fanfarenkonzert auf."""

table = str.maketrans("KFkf", "FKfk")
text = text.translate(table)
print(text)
```

## Regular Expressions

Regular expressions are a powerful tool for searching and manipulating strings based on specific patterns. They can be used for tasks such as validating input, extracting information, or replacing parts of a string. The `re` module in Python provides functions to work with regular expressions.

```python
import re

# Search if a word is in the text:
x = re.search("Kräfte", text)
print(x) # None

x = re.search("Fräkte ", text) # Returns the first match of the pattern in the text

print(type(x),x) # The type is a match object.
print(bool(x)) # You can convert it into a boolean.

# Match objects have some useful methods to get more information about the match:
print(x.span()) # The span method returns the start and end positions of the match in the text. Here it is (0, 7) because "Fluffum" starts at position 0 and ends at position 7 (not included).
print(x.group()) 

# The group method returns the part of the string where there was a match (here trivial, the search string). This will change when the pattern gets more complex.
x = re.search(".unfelkragmente ", text)
print(x.group()) # There is one variable character before "nallen", so it matches "knallen" or "fnallen" in the text.
```

## Pattern Syntax

Regular expressions use a specific syntax to define patterns.

| Character | Description | Example |
|---|---|---|
| `[]` | A set of characters | `"[a-m]"` |
| `\` | Signals a special sequence (can also be used to escape special characters) | `"\d"` |
| `.` | Any character (except newline character) | `"he..o"` |
| `^` | Starts with | `"^hello"` |
| `$` | Ends with | `"planet$"` |
| `*` | Zero or more occurrences | `"he.*o"` |
| `+` | One or more occurrences | `"he.+o"` |
| `?` | Zero or one occurrences | `"he.?o"` |
| `{}` | Exactly the specified number of occurrences | `"he.{2}o"` |
| `\|` | Either or | `"falls\|stays"` |
| `()` | Capture and group |  |

There are some special characters to specify character classes:

| Character | Description | Example |
|---|---|---|
| `\A` | Returns a match if the specified characters are at the beginning of the string | `"\AThe"` |
| `\b` | Returns a match where the specified characters are at the beginning or at the end of a word (the `r` in the beginning makes sure the string is treated as a raw string) | `r"\bain"` / `r"ain\b"` |
| `\B` | Returns a match where the specified characters are present, but NOT at the beginning (or at the end) of a word (the `r` in the beginning makes sure the string is treated as a raw string) | `r"\Bain"` / `r"ain\B"` |
| `\d` | Returns a match where the string contains digits (numbers from 0-9) | `"\d"` |
| `\D` | Returns a match where the string DOES NOT contain digits | `"\D"` |
| `\s` | Returns a match where the string contains a white space character | `"\s"` |
| `\S` | Returns a match where the string DOES NOT contain a white space character | `"\S"` |
| `\w` | Returns a match where the string contains any word characters (characters from a to Z, digits from 0-9, and the underscore `_` character) | `"\w"` |
| `\W` | Returns a match where the string DOES NOT contain any word characters | `"\W"` |
| `\Z` | Returns a match if the specified characters are at the end of the string | `"Spain\Z"` |

You can also define sets of characters:

| Set | Description |
|---|---|
| `[arn]` | Returns a match where one of the specified characters (`a`, `r`, or `n`) is present |
| `[a-n]` | Returns a match for any lower case character, alphabetically between `a` and `n` |
| `[^arn]` | Returns a match for any character EXCEPT `a`, `r`, and `n` |
| `[0123]` | Returns a match where any of the specified digits (`0`, `1`, `2`, or `3`) are present |
| `[0-9]` | Returns a match for any digit between `0` and `9` |
| `[0-5][0-9]` | Returns a match for any two-digit numbers from `00` to `59` |
| `[a-zA-Z]` | Returns a match for any character alphabetically between `a` and `z`, lower case OR upper case |
| `[+]` | In sets, `+`, `*`, `.`, `|`, `()`, `$`, `{}` have no special meaning, so `[+]` means: return a match for any `+` character in the string |

The regex object has some methods to work with regular expressions:

| Function | Description |
|---|---|
| `findall` | Returns a list containing all matches |
| `search` | Returns a Match object if there is a match anywhere in the string |
| `split` | Returns a list where the string has been split at each match |
| `sub` | Replaces one or many matches with a string |
| `match` | Checks for a match only at the beginning of the string. If the whole string matches the pattern |

```python
# Example of a regular expression pattern:
text = """Die Kräfte knallen, die Fäden verflechten sich, ein Klangteppich voller Funkelfragmente entsteht.
Man könnte fast glauben, ein Kakadu führe ein Fanfarenkonzert auf."""

pattern = r"\b\w+en\b" # This pattern matches any word that ends with "um". The \b means a word boundary, \w+ means one or more word characters, and um is the literal string "um". So it will match words like "Fluffum", "schnipsum", "wackelito", etc.
matches = re.findall(pattern, text) # The findall function returns all non-overlapping matches of the pattern in the text as a list of strings.
print(matches)
```

## Example

Here is an example wich is hopefully more interesting than the lorem ipsum examples before. It combines the techniques to analyze a political speech. The rhetorical method "Repetitio" is used to repeat words. It is often used to emphasize a point or to make an idea more memorable. The evil way to use it is to manipulate or deceive. The example I use here is the historical speech from Joseph Goebbels known as "Total War Speech" or "Sportpalast speech". You will finde the german original at [Sportpalast-Rede von Joseph Goebbels – 1000 Schlüsseldokumente](https://www.1000dokumente.de/Dokumente/Sportpalast-Rede_von_Joseph_Goebbels). I choose this speech because it is a well-known historical example. It is not very complex in terms of content. Its only purpose was to act manipulatively as propaganda and to prepare people for the war. Perhaps you know the audience’s reaction, when it broke out into thunderous applause. Not too difficult with an invited audience. 

We want to find out which words are repeated most often in the speech. Because stop words will be the most frequent, we will remove them. The stop words are stored in the `stopwords` corpus from the `nltk` package. Finally, we will visualize the most common words in a bar chart.

Let's see what insights we can get about this speech.

```python
import re
import matplotlib.pyplot as plt
import nltk

from collections import Counter
from nltk.corpus import stopwords

# Read the speech from a text file:
with open("../assets/texts/historical_speech.txt", "r", encoding="utf-8") as f:
    speech = f.read()

# Remove all chars that are not word characters or whitespace using a regular expression.
cleaned = re.sub(r"[^\w\s]", "", speech)
# Splitting the cleaned text into words:
words = cleaned.split()

# Remove stop words with the help of the nltk library:
nltk.download("stopwords")
german_stops = set(stopwords.words("german"))
filtered = [w for w in words if w.lower() not in german_stops]
counter = Counter(filtered)

# Plot the word frequencies:
most_common = counter.most_common(15)
labels, values = zip(*most_common)

plt.figure(figsize=(25, 10))
plt.title("Most common words in the speech")
plt.xlabel("Words")
plt.ylabel("Frequency")
plt.yticks(values)
plt.bar(labels, values)
plt.show()
```

The speech is in German. Here the most common words are containing: People, Germans, danger, "The Fuhrer", front, Bolshevism, war, fight. Can you recognize a tendency? I always find it fascinating how often you can guess the intention of a text just by looking at the most common words.

Stop words are a simple way to eliminate unwanted words. `spacy` is another lib which has a more sophisticated way to filter out unwanted words. It is a powerful library for natural language processing (NLP) in Python. It provides various features for text processing, including tokenization, part-of-speech tagging, named entity recognition, and more.

I will use another dataset for this example. [world-politics-datalab/trump\_presidency\_tweets: This repository includes all of President Trump tweets posted during his presidency.](https://github.com/world-politics-datalab/trump_presidency_tweets?tab=readme-ov-file) contains tweets from Donald Trump during his presidency. We will download the dataset load it with `pandas` and process it with `spacy`.

```python
import urllib.request
import pandas
import spacy
import matplotlib.pyplot as plt
import os

from collections import Counter
from spacy.cli.download import download

url = "https://raw.githubusercontent.com/world-politics-datalab/trump_presidency_tweets/refs/heads/main/trump_presidency_tweets_2017_2020.csv"
download_path = "../temp/trumps_tweets.csv"

if not os.path.exists(download_path):
    os.makedirs(os.path.dirname(download_path), exist_ok=True)
    urllib.request.urlretrieve(url, download_path)
    print(f"Datei heruntergeladen: {download_path}")

df = pandas.read_csv(download_path)
df.head()

model_name = "en_core_web_sm"

try:
    nlp = spacy.load(model_name)
except OSError:
    download(model_name)
    nlp = spacy.load(model_name)

filtered_words = []
texts = df["text"].dropna().astype(str).tolist()

for doc in nlp.pipe(texts, batch_size=200, disable=["parser", "ner"]):
    filtered_words.extend(
        token.lemma_.lower()
        for token in doc
        if token.pos_ in {"NOUN", "ADJ", "PROPN"}
        and token.is_alpha
        and not token.is_stop
    )

counter = Counter(filtered_words)

# Plot the word frequencies:
most_common = counter.most_common(20)
labels, values = zip(*most_common)

plt.figure(figsize=(25, 10))
plt.title("Most common words in Trump's tweets")
plt.xlabel("Words")
plt.ylabel("Frequency")
plt.yticks(values)
plt.bar(labels, values)
plt.show()
```

Big news!. The result is not surprising. "Trump", "people", "great", "America", "country", "time", "democrats", "fake", "job", "border" and "today" are in the most common words.
